# Charts Design

### Monthly Sales Trend

### Key Insights

- Monthly sales show noticeable fluctuations throughout the period rather than a consistent upward trend.
- February recorded the lowest monthly sales, followed by a strong recovery in March and April.
- Sales reached a high point in May before declining in June.
- Sales recovered again during July and August, indicating renewed revenue momentum toward the end of the analysis period.
- The fluctuations suggest that monthly sales performance may be influenced by changes in demand and customer ordering patterns.

```python
df["Order Date"] = pd.to_datetime(df["Order Date"])

df["YearMonth"] = df ["Order Date"].dt.to_period("M").astype(str)

monthly_revenue =df.groupby("YearMonth")["Price (INR)"].sum().reset_index()

plt.figure()
plt.plot(monthly_revenue["YearMonth"], monthly_revenue["Price (INR)"])
plt.xticks(rotation=45)
plt.xlabel("Month")
plt.ylabel("Toatl Sales (INR")
plt.title("Monthly Sales Trend")
plt.tight_layout()
plt.show()
```

### Daily Sales Trend

### Key Insights

- Saturday generated the highest revenue among the seven days.
- Tuesday recorded the lowest revenue.
- Revenue remained relatively consistent across the remaining days, with only moderate variation.
- The higher revenue observed on Saturday suggests stronger customer spending or ordering activity toward the weekend.
- Businesses could consider using weekend-focused promotions and offers to further capitalize on higher Saturday demand.

```python
df["DayName"] = pd.to_datetime(df["Order Date"]).dt.day_name()

daily_revenue =(
    df.groupby("DayName")["Price (INR)"]
      .sum()
      .reindex(["Monday","Tuesday","Wednesday","Thursday","Friday","Saturday","Sunday"])
)
plt.figure(figsize=(10,5))
plt.bar(daily_revenue.index, daily_revenue.values)
plt.title("Daily Revenue Trend (Mon-Sun)")
plt.xlabel("Day")
plt.ylabel("Revenue (INR)")
plt.xticks(rotation=30)

plt.show()
```

### Total Sales by Food Type(Veg vs Non-Veg)

### Key Insights

- Vegetarian food contributed approximately 65.2% of total revenue, making it the dominant food category.
- Non-vegetarian food contributed approximately 34.8% of total revenue.
- The analysis indicates a significantly higher revenue contribution from vegetarian offerings.
- This suggests that vegetarian products represent an important part of the analyzed sales mix and could be considered when planning menu offerings and promotions.
- Food categories were classified using keyword-based rules applied to dish names. Therefore, the Veg vs Non-Veg analysis depends on the accuracy and consistency of dish-name descriptions in the dataset.

```python
non_veg_keywords =[
    "chicken","egg","fish","mutton","prawns","biriyani","kabab","kebab","non-veg","non veg"
]

df["Food Category"] = np.where(
    df["Dish Name"].str.lower().str.contains("|".join(non_veg_keywords), na=False),
    "Non-Veg",
    "Veg"
)
```

```python
food_revenue = (
    df.groupby("Food Category")["Price (INR)"]
    .sum()
    .reset_index()
)
```

```python
fig = px.pie(
    food_revenue,
    values="Price (INR)",
    names="Food Category",
    hole=0.5,
    title="Revenue Contribution: Veg vs Non-Veg",
)

fig.update_traces(
    textinfo="percent+label",
    pull=[0.05,0]
)

fig.update_layout(
    height=500,
    margin=dict(t=60, b=40, l=40, r=40)
)

fig.show()
```

### Revenue by State

### Key Insights

- Karnataka generated the highest revenue among all states in the dataset.
- Karnataka's revenue is substantially higher than that of the other states, indicating strong market performance.
- Uttar Pradesh, Telangana and Maharashtra were the next strongest contributors.
- Revenue is distributed across a large number of states, but the top-performing states contribute disproportionately to overall sales.
- The variation in state-level revenue highlights potential differences in market size, customer demand and business presence across regions.

```python
fig =px.bar(
    df.groupby("State", as_index=False)["Price (INR)"].sum()
    .sort_values("Price (INR)", ascending=False),
    x="Price (INR)",
    y="State",
    orientation="h",
    title="Revenue by State (INR)"
)

fig.update_layout(height=600, yaxis=dict(autorange="reversed"))
fig.show()
```

### Quarterly Performance Summary

### Key Insights

- Q2 recorded the highest sales at approximately ₹1.99 crore, slightly higher than Q1 at approximately ₹1.97 crore.
- Q2 also recorded the highest number of orders, with 74,163 orders.
- Q3 shows lower sales and order volume compared with Q1 and Q2; however, Q3 should be interpreted cautiously because the dataset only covers January to August 2025 and therefore does not represent a complete quarter.
- Average ratings remained stable at approximately 4.34 across all three quarters.
- The stable average rating alongside fluctuations in sales suggests that changes in sales performance were not accompanied by significant changes in average customer ratings.

```python
df["Order_Date"] = pd.to_datetime(df["Order Date"])
df["Quarter"]= df["Order_Date"].dt.to_period("Q").astype(str)
quarterly_summary = (
    df.groupby("Quarter", as_index=False)
      .agg(
          Total_Sales=("Price (INR)", "sum"),
          Avg_Rating=("Rating", "mean"),
          Total_Orders=("Order_Date", "count")
      )
      .sort_values("Quarter")
)
quarterly_summary["Total_Sales"]=quarterly_summary["Total_Sales"].round(0)
quarterly_summary["Avg_Rating"]=quarterly_summary["Avg_Rating"].round(2)

quarterly_summary
```

### Top 5 Cities by Sales

### Key Insights

- Bengaluru generated the highest sales among the top five cities analyzed.
- Lucknow ranked second, followed by Hyderabad, Mumbai and New Delhi.
- Bengaluru's sales were substantially higher than those of the other top-performing cities.
- Hyderabad and Mumbai showed relatively similar sales performance.
- The concentration of sales in these cities highlights important high-performing urban markets within the dataset.

```python
top_5_cities = (
    df.groupby("City")["Price (INR)"]
      .sum()
      .nlargest(5)
    .sort_values()
      .reset_index()
)

fig = px.bar(
    top_5_cities,
    x="Price (INR)",
    y="City",
    orientation="h",
    title="Top 5 Cities by Sales (INR)",
    color_discrete_sequence=["red"]
)

fig.show()
```

### Weekly Revenue Trend

### Key Insights

- Weekly revenue remained relatively stable throughout the analyzed period.
- Most weeks generated revenue within a relatively narrow range.
- The highest weekly revenue was observed around the middle of the period, while some earlier weeks recorded comparatively lower revenue.
- The relatively consistent weekly performance suggests that revenue was not highly volatile on a week-to-week basis.
- This stability can support more predictable operational and inventory planning.

```python
df["Week"] = pd.to_datetime(df["Order Date"]).dt.isocalendar().week

weekly_revenue = (
    df.groupby("Week")["Price (INR)"]
    .sum()
    .reset_index()
)

fig = px.bar(
    weekly_revenue,
    x="Week",
    y="Price (INR)",
    title="Weekly Revenue Trend",
    text_auto=True,
)
fig.show()
```
